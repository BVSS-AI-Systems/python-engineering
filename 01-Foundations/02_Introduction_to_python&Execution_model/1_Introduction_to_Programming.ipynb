# Part 1 — What does a computer actually do?

Strip away the screens, apps and logos, and **every** computer (laptop, phone, game console, smart watch, a car's computer) does the same four things:

| Job | What it means | Hardware that does it |
|---|---|---|
| **Input** | take information in | keyboard, mouse, touchscreen, microphone, camera |
| **Process** | do something with it (calculate, compare, decide) | **CPU**, the Central Processing Unit |
| **Output** | send the result out | screen, speakers, printer |
| **Store** | remember things | **RAM** (short-term) and **storage** (long-term) |



### The main parts, in plain words

**CPU (Central Processing Unit): the brain**
- Does all the calculating and decision-making.
- It is *extremely* fast (billions of tiny steps per second) but *extremely* simple. Each step is something like "add these two numbers" or "is this number bigger than that one?"
- It has **no ideas of its own**. It only does what instructions tell it.

**RAM (Random Access Memory): short-term memory**
- Holds whatever the computer is working on *right now*: the program that's running and the data it's using.
- Very fast, but **temporary**: everything in RAM disappears when the power goes off. (That's why unsaved work is lost in a power cut.)

**Storage (SSD / hard disk): long-term memory**
- Where your files, photos, apps and Python itself live permanently.
- Much slower than RAM, but it **keeps data when the power is off**.
- When you open a program, the computer **copies it from storage into RAM**, and then the CPU runs it from there.

### An analogy that helps: the kitchen



- The **chef (CPU)** is incredibly fast, but just stands there unless there's a recipe.
- The **countertop (RAM)** is where the ingredients being used *right now* sit. It's small and gets cleared at the end of the day.
- The **fridge/pantry (storage)** holds everything long-term. Fetching from it is slower.
- The **recipe (program)** tells the chef exactly what to do, step by step.

**Programming = writing the recipe.**

### Everything is 0s and 1s

Inside the computer there are no letters, colours or sounds. There are only billions of microscopic switches called **transistors**, and each one is either **OFF (0)** or **ON (1)**. One 0-or-1 is called a **bit**; 8 bits make a **byte**.

So how does a computer store the letter `A`? It agrees on a code: `A` is the number **65**, and 65 written in binary (base 2) is **01000001**. Photos, music and videos work the same way: lots and lots of numbers, which are lots and lots of bits.

# Part 2 — What is a program?

> **A program is a list of exact instructions that a computer follows, in order.**

That's it. A web browser, WhatsApp, a game, a calculator: each is a (very long) list of instructions written by programmers.

Here's a tiny program, first in plain English, then in Python:



### Three rules of how computers follow instructions

**1. Order matters: top to bottom.**
The computer starts at the first instruction and goes down, one at a time. It can't "jump ahead" and use something you haven't created yet.

**2. Exact means exact.**
A human understands "make me a sandwich". A computer needs every tiny step spelled out. If you leave out a step, it doesn't guess. It does the wrong thing, or stops with an error.

**3. It does what you *wrote*, not what you *meant*.**



A mistake in a program is called a **bug**, and finding and fixing bugs is called **debugging**. *Every* programmer writes bugs every day. It's a normal part of the job, not a sign that you're bad at it.

### See the rules in action

**Order matters.** The cell below tries to *use* `total` before it is *created*. Run it and you'll see an error. That's expected!

````python
print(total)    # line 1: tries to show 'total'... but it doesn't exist yet!
total = 10      # line 2: 'total' is only created here
````

Python read line 1 first, didn't know what `total` was, and stopped with a **`NameError`**. Now the same two lines in the **correct order**:

````python
total = 10
print(total)
````

**Exact means exact.** Python is picky about spelling and capital letters. `print` works, but `Print` is a different word to Python:

````python
Print("Hello")   # capital P: Python doesn't know this word
````

**Top to bottom, and it stops at the first error.** Once Python hits an error, nothing after it in that cell runs.



````python
print("one")
print("two")
print(10 / 0)      # you can't divide by zero, so Python stops here
print("three")     # this line never runs
````

### How to read an error message (don't panic!)

Red text isn't the computer being angry. It's Python **telling you exactly what went wrong**. Read the **last line** first:

```
ZeroDivisionError: division by zero
^^^^^^^^^^^^^^^^^  ^^^^^^^^^^^^^^^^
  type of error      what happened
```

Look a little higher and Python also shows an arrow `---->` pointing at **which line** caused it. So every error message answers *what* went wrong and *where*.

> **Key takeaway (Part 2):** A program is a list of **exact** instructions run **top to bottom**. The computer does what you *wrote*, not what you *meant*. Mistakes are called **bugs**, and error messages are clues that help you fix them.

# Part 3 — Why do programming languages exist?

### The problem

The CPU only understands **machine code**: instructions written as raw 0s and 1s, like

```
10111000 00000010 00000000 00000000 00000000
```

Writing a whole app like that would be painfully slow and almost impossible to read or fix. Early programmers really did work this way, flipping switches and punching holes in cards!

### The solution: languages closer to human words

Over the decades people invented languages that sit **higher up**, closer to how humans think. Each level gets translated down to the one below until it reaches machine code.



All four rows do the same thing: add 2 and 3. Look how much easier the Python version is to read. Writing simple, readable code for humans and letting a tool handle the machine details is called **abstraction**.

### Someone has to translate

Since the CPU only speaks machine code, your Python must be **translated**. There are two classic approaches:



| | **Compiler** (C, C++, Go, Rust) | **Interpreter** (Python, JavaScript) |
|---|---|---|
| When does translation happen? | all at once, **before** running | **while** running |
| What do you get? | a separate program file (e.g. `.exe`) | results straight away |
| Speed | usually faster to run | usually slower to run |
| Trying ideas quickly | write → compile → run | write → run ✔ |
| Analogy | translating a whole book, then publishing it | a live interpreter at a meeting, sentence by sentence |

**Python uses an interpreter**, a program called `python` that you installed. When you press Shift + Enter, Jupyter hands your code to the Python interpreter, which translates it and makes the CPU carry it out.

> 🔎 *Curious detail (you'll see this later in the course):* Python actually does a quick "half-translation" into something called **bytecode** first, then runs that. For now, "Python is an interpreted language" is the right mental model.

### Why Python, of all languages?

There are hundreds of languages. Python is one of the most popular in the world, especially for beginners:

- **Reads almost like English:** `if age >= 18: print("adult")`
- **Less "ceremony":** printing one line in Python is 1 line. In Java it takes about 5.
- **Interactive:** run a line and see the result immediately, like you're doing in Jupyter.
- **Used everywhere:** websites (Instagram, YouTube's backend), data science, AI and machine learning, automation, cloud (AWS tools!), science and finance.
- **Huge community:** almost any question you have has already been answered online.
