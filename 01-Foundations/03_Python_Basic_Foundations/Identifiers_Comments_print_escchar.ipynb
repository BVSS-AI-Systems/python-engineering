# What is an Identifier in Python?

An identifier is a name used to identify and refer to program elements such as variables, functions, classes, modules, and objects. It acts as a label that allows Python to recognize and access these elements throughout a program.

Identifiers make code readable, meaningful, and easier to maintain by giving descriptive names to different components of a program.

---

# Rules for Naming Identifiers

### 1. Allowed Characters

Identifiers can contain:

- Letters (A-Z, a-z)
- Digits (0-9)
- Underscores (_)

---

### 2. Starting Character

An identifier must begin with:

- A letter, or
- An underscore (_)

It cannot begin with a digit.

---

### 3. No Special Characters

Identifiers cannot contain special characters such as:

- @
- #
- $
- %
- -
- !

Only letters, digits, and underscores are permitted.

---

### 4. Keywords Cannot Be Used

Reserved Python keywords have special meanings and cannot be used as identifiers.

Examples include words used for control flow, class definitions, exception handling, and other language constructs.

---

### 5. Case Sensitivity

Python identifiers are case-sensitive.

Names that differ only in letter case are treated as completely different identifiers.

---

# Best Practices for Naming Identifiers

- Use meaningful and descriptive names.
- Choose names that clearly represent their purpose.
- Use underscores to improve readability when names contain multiple words.
- Follow a consistent naming style throughout a project.
- Avoid overly short or cryptic names unless their purpose is obvious.

---

# Importance of Identifiers

Identifiers are fundamental to programming because they:

- Improve code readability.
- Make programs easier to understand.
- Help organize and manage program components.
- Enable developers to write maintainable and reusable code.

# Identifier Naming Conventions in Python

Naming conventions are standardized styles used for naming identifiers such as variables, functions, classes, modules, and constants. Following consistent naming conventions improves code readability, maintainability, and collaboration among developers.

---

# 1. snake_case

In snake_case, words are written in lowercase and separated by underscores (_).

Examples:

- student_name
- total_marks
- employee_salary
- calculate_average

This is the most commonly used naming convention in Python.

### Typically Used For

- Variables
- Functions
- Modules

---

# 2. camelCase

In camelCase, the first word starts with a lowercase letter, and each subsequent word begins with an uppercase letter.

Examples:

- studentName
- totalMarks
- employeeSalary
- calculateAverage

This convention is commonly used in languages such as Java and JavaScript but is generally not preferred in Python.

---

# 3. PascalCase (CapWords)

In PascalCase, every word begins with an uppercase letter.

Examples:

- Student
- BankAccount
- EmployeeDetails
- CustomerRecord

### Typically Used For

- Class Names

Python's style guide recommends PascalCase for naming classes.

---

# 4. UPPER_CASE

In UPPER_CASE, all letters are uppercase and multiple words are separated by underscores.

Examples:

- PI
- MAX_SIZE
- DATABASE_URL
- DEFAULT_TIMEOUT

### Typically Used For

- Constants
- Configuration Values

These names indicate values that should not be changed during program execution.

---

# 5. Leading Underscore Convention

Identifiers beginning with a single underscore indicate that they are intended for internal use.

Examples:

- _name
- _balance
- _helper_function

This is a naming convention and does not enforce strict access restrictions.

---

# Python (PEP 8) Naming Recommendations

| Identifier Type | Recommended Style |
|----------------|-------------------|
| Variables | snake_case |
| Functions | snake_case |
| Modules | snake_case |
| Packages | lowercase |
| Classes | PascalCase |
| Constants | UPPER_CASE |
| Internal Members | _leading_underscore |

---

# Why Follow Naming Conventions?

Using consistent naming conventions:

- Improves code readability.
- Makes programs easier to maintain.
- Enhances collaboration among developers.
- Helps identify the purpose of an identifier quickly.
- Follows industry-standard Python practices.

---

Python uses different naming conventions for different types of identifiers. The most commonly used convention is snake_case for variables and functions, PascalCase for classes, and UPPER_CASE for constants. Following these conventions ensures clean, readable, and professional Python code.

# The print() Function

The `print()` function is a built-in Python function used to display output on the console or terminal.

It is one of the most frequently used functions in Python and is commonly used for:

- Displaying messages
- Showing variable values
- Debugging programs
- Presenting results to users

---

## Purpose of print()

The primary purpose of the `print()` function is to send information from a program to the standard output device, typically the console.

---

## Characteristics of print()

- It is a built-in function.
- It does not require importing any module.
- It can display text, numbers, variables, and objects.
- It automatically converts values into a readable format before displaying them.
- It supports customization through optional parameters.

---

## Parameters of print()

| Parameter | Description |
|------------|-------------|
| objects | Values to be displayed |
| sep | Separator used between multiple values |
| end | String appended after the output |
| file | Output destination |
| flush | Controls output buffering |

---

## Separator (sep)

The separator determines how multiple values are separated when displayed.

By default, Python uses a single space as the separator.

Common separators include:

- Space
- Comma
- Hyphen
- Pipe Symbol
- Colon

---

## End Character (end)

The end parameter determines what is appended after the output is displayed.

By default, Python appends a newline character, causing the next output to appear on a new line.

The end parameter can be modified to control the formatting of the output.

---

## Working with Multiple Values

The `print()` function can display multiple values in a single statement.

When multiple values are supplied, Python automatically inserts the separator between them.

---

## How Python Executes print()

When Python encounters a `print()` call:

1. The interpreter recognizes `print` as a built-in function.
2. The supplied values are evaluated.
3. The values are converted to their string representation.
4. The output is sent to the console.
5. The specified separator and ending characters are applied.

---

## print() is Not a Keyword

Although it appears frequently in Python programs, `print` is not a keyword.

It is a built-in function provided by Python's built-in namespace.

This means Python treats `print` as a function rather than a language keyword.

---

## Built-in Functions vs Keywords

| Built-in Functions | Keywords |
|-------------------|------------|
| Perform operations | Define Python syntax and structure |
| Can be called like functions | Reserved words |
| Examples: print(), len(), type() | Examples: if, for, while, class |
| Provided by Python's built-in namespace | Part of Python language grammar |

---

## Key Takeaways

- Built-in functions are predefined functions provided by Python.
- They can be used without importing any modules.
- `print()` is one of the most commonly used built-in functions.
- The primary purpose of `print()` is to display output.
- `print()` supports customization through parameters such as `sep` and `end`.
- `print()` is a built-in function, not a Python keyword.

# Escape Sequences in Python

Escape sequences are special character combinations that begin with a backslash (`\`). They are used to represent characters that cannot be typed directly or have special meanings.

## Common Escape Sequences

| Escape Sequence | Description |
|----------------|-------------|
| `\n` | New line |
| `\t` | Horizontal tab |
| `\\` | Backslash character |
| `\'` | Single quote |
| `\"` | Double quote |
| `\b` | Backspace |
| `\r` | Carriage return |

## New Line (`\n`)

Moves the cursor to the next line, allowing text to be displayed on multiple lines.

## Tab (`\t`)

Inserts a horizontal tab space between text, useful for alignment and formatting output.
