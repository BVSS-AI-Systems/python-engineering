## 1. What is a Variable?
 
A **variable** is a **name** that refers to a **value** (an object) stored in the computer's memory.
 
Think of it as a **label (name tag)** stuck on a box, not the box itself.


````python
age = 25
````

# Everything in Python is an Object

Python is an object-oriented programming language in which every value is treated as an object.

An object is an entity that contains:

- Identity
- Type
- Value

Variables do not store values directly. Instead, they store references to objects in memory.

```python
age = 25
```
 
What happens behind the scenes:
1. Python creates an **object** `25` (of type `int`) somewhere in memory.
2. Python creates the **name** `age`.
3. The name `age` is **bound** (points) to that object.
> **Key idea:** In Python, variables don't *store* values directly — they *reference* objects in memory.
 
```
  age  ───────►  [ 25 ]   (int object in memory)
```
 
---

Examples of objects in Python include:

- Integers
- Floating-point numbers
- Strings
- Lists
- Tuples
- Dictionaries
- Sets
- Functions
- Classes

Because everything is an object, every value can have associated attributes and methods that define its behavior.

This object-oriented design makes Python flexible, powerful, and consistent across different data types.

## 2. Why Do We Need Variables?
 
- **Store data** to use later (user input, results of calculations)
- **Reuse values** without retyping them
- **Make code readable** — `tax_rate = 0.18` is clearer than a bare `0.18`
- **Change values easily** — update one place, not everywhere

## 3. Creating (Assigning) Variables
 
Python uses the **assignment operator** `=`.
 
```python
name = "Ravi"        # string
age = 25             # integer
height = 5.9         # float
is_student = True    # boolean
```
 
- Left side → **variable name**
- Right side → **value / expression**
- `=` means "assign", **not** "equal to" (equality check is `==`)
### No declaration needed
Unlike C or Java, you don't declare the type first:
 
```c
// C / Java
int age = 25;
```
```python
# Python
age = 25
```
 
A variable is created the **moment you first assign** a value to it.
 

````python
name = 'Sai Vardhan'
age = 27
height = 5.10
is_student = True
````

## 4. Dynamic Typing
 
Python is **dynamically typed** — the type belongs to the **value**, not the variable name. The same name can point to different types over time.
 
```python
x = 10        # x refers to an int
x = "hello"   # now x refers to a str
x = 3.14      # now x refers to a float
```
 
Check the type with `type()`:
 
```python
x = 10
print(type(x))   # <class 'int'>
 
x = "hello"
print(type(x))   # <class 'str'>
```
 
> Python is also **strongly typed**: it won't silently mix incompatible types.
> `"5" + 5` → `TypeError`. You must convert: `int("5") + 5` → `10`.
 

````python
x = 10
print(f"Value in x : {x} and its type is {type(x)}")
x = "Hello"
print(f"Value in x : {x} and its type is {type(x)}")
````

## 5. Variable Naming Rules (Must Follow)
 
| Rule | Valid ✅ | Invalid ❌ |
|------|---------|-----------|
| Must start with a letter or underscore `_` | `name`, `_count` | `1name` |
| Can contain letters, digits, underscores only | `user_1` | `user-1`, `user 1`, `user@1` |
| Case-sensitive | `Age` and `age` are **different** | — |
| Cannot be a **keyword** | `my_class` | `class`, `if`, `for` |
 

 

````python
age = 20
Age = 30
AGE = 40
print(age, Age, AGE)   # 20 30 40  → three different variables
````

## 7. Variables Are References — `id()` and `is`
 
Every object in memory has a unique identity. `id()` shows it.
 
```python
a = 100
b = a
print(id(a))
print(id(b))   # same id → both names point to the SAME object
print(a is b)  # True
```
 
```
  a ──┐
      ├──►  [ 100 ]
  b ──┘
```

````python
a = 100
a = b
print(f"The id of value {a} is {id(a)}")
print(f"The id of value {b} is {id(b)}")
````

### `==` vs `is`
| Operator | Checks |
|----------|--------|
| `==` | Do they have the **same value**? |
| `is` | Are they the **same object** in memory? |
 
```python
x = [1, 2, 3]
y = [1, 2, 3]
print(x == y)  # True  (same values)
print(x is y)  # False (two different list objects)
```
 
> Use `is` mainly for checking `None`: `if value is None:`
 

````python
a = [10, 20, 30]
b = [10, 20, 30]
print(f"The id of value {a} is {id(a)}")
print(f"The id of value {b} is {id(b)}")
print(a == b)
print(a is b)
````

## 8. Mutable vs Immutable Objects (Very Important)
 
| Immutable (cannot change in place) | Mutable (can change in place) |
|---|---|
| `int`, `float`, `bool`, `str`, `tuple`, `frozenset` | `list`, `dict`, `set` |
 
### Immutable example
```python
a = 10
b = a
a = a + 5
print(a)  # 15
print(b)  # 10  → b is unaffected
```
`a + 5` creates a **new** object `15`; `a` is re-pointed to it. `b` still points to `10`.

````python
a = 10
b = a
print(f"before addition and updating a:{a},{id(a)}")
print(f"before addition and updating b:{b},{id(b)}")
a+=5
print(f"After addition and updating a:{a},{id(a)}")
print(f"After addition and updating b:{b},{id(b)}")
````

 
### Mutable example (common beginner surprise!)
```python
list1 = [1, 2, 3]
list2 = list1          # NOT a copy — same list, two names
list2.append(4)
print(list1)  # [1, 2, 3, 4]  😮
print(list2)  # [1, 2, 3, 4]
```
 
```
  list1 ──┐
          ├──►  [1, 2, 3, 4]
  list2 ──┘
```
 

````python
a = [10, 20, 30, 40]
b = a
print(f"before addition and updating a:{a},{id(a)}")
print(f"before addition and updating b:{b},{id(b)}")
a.append(50)
print(f"After addition and updating a:{a},{id(a)}")
print(f"After addition and updating b:{b},{id(b)}")
````

## 9. Deleting Variables — `del`
 
```python
city = "Hyderabad"
print(city)
del city
print(city)   # NameError: name 'city' is not defined
```
 
`del` removes the **name**, not necessarily the object. When no names refer to an object anymore, Python's **garbage collector** frees the memory automatically.
 
---

````python
my_name = "Venkata Babburi"
print(my_name)
del my_name
print(my_name)
````

## 10. Common Errors with Variables
 
| Error | Cause | Example |
|-------|-------|---------|
| `NameError` | Using a variable before assigning it (or typo) | `print(totl)` |
| `SyntaxError` | Invalid name | `2value = 5`, `my-var = 5` |
| `TypeError` | Mixing incompatible types | `"Age: " + 25` |
| `ValueError` | Unpacking count mismatch | `a, b = 1, 2, 3` |
 
Fix for the `TypeError` example:
```python
age = 25
print("Age: " + str(age))   # convert to str
print(f"Age: {age}")        # ✅ f-string (best)
```
