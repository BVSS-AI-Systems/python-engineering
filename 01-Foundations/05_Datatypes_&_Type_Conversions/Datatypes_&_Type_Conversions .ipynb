## 1. What is a Data Type?

A **data type** tells Python:
1. **What kind of value** something is (a number, text, a list...)
2. **How it is stored** in memory
3. **What operations** you can do with it

The same symbol (`+`) behaves differently depending on the type:

````python
print(10 + 5)     # numbers are added
print("Hi"+ "!")  # strings are joined
print("Hi" * 3)   # strings are repeated
````

**⚠️ Error demo:** you can't add a number and text.

````python
#print(10 + "5")    #TypeError: unsupported operand type(s) for +: 'int' and 'str'
print(10 + int("5"))   # convert first
````

> In Python **every value is an object**, and **the type belongs to the object, not the variable**.

# Python int Internals - Number Systems and Storage

## Important Concept

**Number System (Representation) ≠ Internal Storage**

When we write numbers in Python, we can use different number systems:

- Decimal (Base 10)
- Binary (Base 2)
- Octal (Base 8)
- Hexadecimal (Base 16)

However, Python does **not** store them differently internally. After parsing the source code, all of them become a single `int` object (`PyLongObject`).

---

## Different Representations

| Number System | Syntax | Value |
|--------------|---------|--------|
| Decimal | `25` | 25 |
| Binary | `0b11001` | 25 |
| Octal | `0o31` | 25 |
| Hexadecimal | `0x19` | 25 |

Although the representations are different, they all produce the same integer value.

---

## Compilation Stage

Python Source Code:

25
0b11001
0o31
0x19

### Lexical Analysis

Python tokenizer recognizes:

- Decimal literal
- Binary literal
- Octal literal
- Hexadecimal literal

### Parsing

Parser creates the numeric literal nodes in the AST.

### Object Creation

The interpreter converts every numeric literal into a decimal integer value and creates a `PyLongObject`.

---

## Internal Storage (PyLongObject)

Conceptually:

PyLongObject
│
├── Object Header
├── Sign Information
└── Digits Array

Python stores integers using an array of digits.

For a small number:

25

Internally:

Sign = Positive
Digits = [25]

For a larger number:

12345678901234567890

Internally:

Sign = Positive
Digits = [chunks of the number]

---

## Example

### Binary Input

0b11001

Conversion:

11001₂

↓

16 + 8 + 1

↓

25₁₀

Stored Internally:

PyLongObject
Sign = Positive
Digits = [25]

---

## Hexadecimal Input

0x19

Conversion:

1 × 16 + 9

↓

25₁₀

Stored Internally:

PyLongObject
Sign = Positive
Digits = [25]

---

## What Python Actually Stores

Python does NOT remember:

- "This number came from binary"
- "This number came from octal"
- "This number came from hexadecimal"

It only remembers the final integer value.

Example:

25
0b11001
0o31
0x19

All become the same integer object value:

25

---

## Internal Digits

CPython stores large integers in chunks.

On modern 64-bit CPython:

- Each chunk stores 30 bits.
- Multiple chunks are used for very large integers.
- Memory grows automatically as needed.

Example:

10**100

Internally:

PyLongObject
│
├── Sign
└── Digits = [chunk1, chunk2, chunk3, ...]

This is why Python integers have arbitrary precision.

---

## Key Takeaways

1. Python supports Decimal, Binary, Octal, and Hexadecimal literals.
2. Number systems are only source-code representations.
3. All representations become a `PyLongObject`.
4. Python stores the integer value, not the original notation.
5. Large integers are stored as an array of chunks.
6. The practical limit of `int` is available memory.
7. Python integers are arbitrary-precision integers.

# Why Does `0.1 + 0.2` Produce `0.30000000000000004`?

## The Problem

```python
print(0.1 + 0.2)
```

Output:

```text
0.30000000000000004
```

Expected:

```text
0.3
```

Actual:

```text
0.30000000000000004
```

---

## Understanding the Root Cause

The issue is **not** with Python.

It arises because computers store floating-point numbers using the **IEEE 754 Floating-Point Standard**.

Computers work with **binary (base 2)**, while humans usually work with **decimal (base 10)**.

---

## Decimal vs Binary Representation

Some numbers can be represented exactly in binary.

Example:

```text
0.5
```

Binary:

```text
0.1₂
```

Exact representation ✅

---

Some numbers cannot be represented exactly.

Example:

```text
0.1
```

Binary:

```text
0.0001100110011001100110011...
```

The pattern repeats forever.

---

Similarly:

```text
0.2
```

becomes

```text
0.001100110011001100110011...
```

which also repeats forever.

---

## The Storage Limitation

A Python float follows the IEEE 754 double-precision format.

A float gets only:

```text
64 bits
```

of storage.

Since the binary expansion of `0.1` never ends, Python must store an approximation.

---

## What Python Actually Stores

Conceptually:

```text
0.1
```

is stored approximately as:

```text
0.10000000000000000555...
```

and

```text
0.2
```

is stored approximately as:

```text
0.20000000000000001110...
```

---

## Internal Addition

Python adds the stored approximations:

```text
0.10000000000000000555...
+
0.20000000000000001110...
--------------------------------
0.30000000000000004440...
```

Displayed as:

```text
0.30000000000000004
```

---

## Why Does `round()` Fix It?

```python
round(0.1 + 0.2, 2)
```

Output:

```text
0.3
```

Explanation:

```text
0.30000000000000004
```

rounded to 2 decimal places becomes:

```text
0.30
```

which is displayed as:

```text
0.3
```

---

## IEEE 754 Double Precision Layout

A floating-point number occupies 64 bits:

```text
+------+-----------+----------------------+
| Sign | Exponent  | Fraction (Mantissa)  |
+------+-----------+----------------------+
 1 bit   11 bits          52 bits
```

### Sign Bit

Stores whether the number is positive or negative.

```text
0 = Positive
1 = Negative
```

---

### Exponent

Stores the power of 2.

Example:

```text
1.01 × 2²
```

Exponent:

```text
2
```

---

### Fraction (Mantissa)

Stores the significant digits of the number.

Example:

```text
1.01 × 2²
```

Fraction stores:

```text
01
```

(The leading 1 is implied and not stored.)

---

## Float Storage Path

```text
0.1
 ↓
Convert to Binary
 ↓
0.00011001100110011...
 ↓
Infinite Repetition
 ↓
Truncate to Available Bits
 ↓
Approximation Stored
 ↓
Used in Calculations
```

---

## Why Integers Don't Have This Problem

### Integer Example

```python
x = 230
```

Python stores:

```text
PyLongObject

ob_digit = [230]
```

The value is stored exactly.

---

### Float Example

```python
x = 0.1
```

Python stores:

```text
PyFloatObject

Sign
Exponent
Fraction
```

Only an approximation of `0.1` can be stored.

---

## Key Takeaways

- Python integers are stored exactly.
- Python floats use IEEE 754 double precision.
- Most decimal fractions cannot be represented exactly in binary.
- `0.1` and `0.2` are stored as approximations.
- Adding approximations can produce tiny errors.
- `round()` only hides the displayed error; it does not change why it occurs.
- Floating-point precision issues exist in almost every programming language, not just Python.

---

## Summary

```text
0.1 + 0.2
```

does not equal an exact binary representation of:

```text
0.3
```

because:

```text
0.1 → approximation
0.2 → approximation
```

and their sum becomes:

```text
0.30000000000000004
```

This behavior is a consequence of the IEEE 754 floating-point representation used by modern computers.

## 5. Boolean Type: `bool`

Only two values: `True` and `False` (capital letters).
### Truthy and Falsy values

| Falsy (count as `False`) | Truthy (count as `True`) |
|---|---|
| `False`, `None` | `True` |
| `0`, `0.0`, `0j` | any non-zero number |
| `""` (empty string) | any non-empty string, even `" "` or `"0"` |
| `[]`, `()`, `{}`, `set()`, `range(0)` | any non-empty collection |

````python
is_logged_in = True
print(is_logged_in)
print(int(is_logged_in))
print(True + True)   # 2 → bool is a subtype of int
````

````python
for value in [0, '', [], None, "0", ' ', [0], [0.001]]:
    print(repr(value),   "->", bool(value))
````

## 7. Sequence Types

### `list`: ordered, changeable, allows duplicates and mixed types

````python
bio_data = ["Sai", {"age":25}, ["Python", "ml"], 180]
bio_data[0] = "Venkata Babburi"
bio_data.append("Chowdary")
print(bio_data)
````

### `tuple`: ordered, **unchangeable**, allows duplicates

Use tuples for fixed data (coordinates, RGB colours, a database row).

````python
point = (10,20)
print(point[0])
# point[0] = 5     # ⚠️ TypeError (immutable)
````

````python
single = (5,)      # a one-item tuple NEEDS a comma
not_tuple = (5)    # this is just int 5
print(type(single), type(not_tuple))
````

### `range`: a memory-efficient sequence of numbers

It doesn't store every number. It calculates them when needed.

````python
print(list(range(5)))           # 0 to 4
print(list(range(2, 10, 2)))    # start, stop, step
````

## 8. Mapping Type: `dict`

Stores **key: value** pairs. You look values up by key, not by position.

- Keys must be **unique** and **immutable** (str, int, tuple...)
- Values can be anything
- Since Python 3.7, dicts keep **insertion order**

````python
student = {
    "name": "Ravi",
    "age": 25,
    "courses": ["Python", "AWS"]
}
print(student["name"])
student["age"] = 26            # update
student["city"] = "Dallas"     # add
print(student.get("phone"))    # None (no error if missing)
print(student)
````

## 9. Set Types

### `set`: unordered, **unique** items, changeable

````python
nums = {1, 2, 2, 3, 3, 3}
print(nums)          # duplicates removed
nums.add(4)
print(3 in nums)     # fast membership check
````

⚠️ `{}` creates an **empty dict**, not a set:

Set maths:

````python
a = {1, 2, 3}
b = {3, 4, 5}
print(a | b)   # union
print(a & b)   # intersection
print(a - b)   # difference
````

### `frozenset`: an immutable set

````python
fs = frozenset([1, 2, 3])
print(fs)
# fs.add(4)   # ⚠️ AttributeError
````

## 12. Type Conversion (Type Casting)

### Implicit: Python converts automatically

````python
x = 5 
print(x, type(x)) 
x += 5.3 
print(x, type(x))    # int promoted to float 
````

### Explicit: you convert using functions

| Function | Example | Result |
|----------|---------|--------|
| `int()` | `int("42")`, `int(3.9)` | `42`, `3` (cuts off the decimal, no rounding) |
| `float()` | `float("3.5")` | `3.5` |
| `str()` | `str(100)` | `'100'` |
| `bool()` | `bool(0)` | `False` |
| `list()` | `list("abc")` | `['a', 'b', 'c']` |
| `tuple()` | `tuple([1, 2])` | `(1, 2)` |
| `set()` | `set([1, 1, 2])` | `{1, 2}` |
| `dict()` | `dict([("a", 1)])` | `{'a': 1}` |

````python
print(int("42"), int(3.9))
print(float("3.5"), str(100), bool(0))
````

````python
print(tuple([1,2]))
````

````python
print(set([1,2,2]))
````

````python
print(dict([("a", 1)]))
````
