# Python Expressions and Operators

## 1. What is an Expression?

An **expression** is any piece of code that Python can **evaluate to produce a value**.

```
   10   +   5        →  15
   ──   ─   ─
operand │ operand
     operator
```

- **Operands:** the values being worked on (`10`, `5`, variables...)
- **Operator:** the symbol that does the work (`+`, `>`, `and`...)

Examples of expressions:

````python
x = 4
print(10 + 5)            # arithmetic expression
print(x * 2 > 7)         # comparison expression → True/False
print("Py" + "thon")     # string expression
print(len("hello"))      # a function call is an expression too
print(x)                 # a single variable is an expression
````

### Expression vs Statement

| Expression | Statement |
|-----------|-----------|
| **Produces a value** | **Performs an action** |
| `3 + 4`, `x > 5`, `len(s)` | `x = 5`, `if ...:`, `for ...:`, `import math` |
| Can be printed / assigned | Can't be printed / assigned |

> `x = 3 + 4` is a **statement** that contains the **expression** `3 + 4`.

In Jupyter, if the **last line** of a cell is an expression, its value is shown automatically:

````python
a = 7
b = 6
a * 6
````

## 2. Types of Operators in Python

| Category | Operators |
|----------|-----------|
| Arithmetic | `+` `-` `*` `/` `//` `%` `**` |
| Comparison (relational) | `==` `!=` `>` `<` `>=` `<=` |
| Logical | `and` `or` `not` |
| Assignment | `=` `+=` `-=` `*=` `/=` `//=` `%=` `**=` `:=` ... |
| Identity | `is` `is not` |
| Membership | `in` `not in` |
| Bitwise | `&` `\|` `^` `~` `<<` `>>` |
| Conditional (ternary) | `x if condition else y` |

Operators can also be grouped by **how many operands** they take:
- **Unary**, one operand: `-x`, `not x`, `~x`
- **Binary**, two operands: `a + b`, `a > b`
- **Ternary**, three parts: `a if cond else b`

## 3. Arithmetic Operators

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Addition | `7 + 2` | `9` |
| `-` | Subtraction | `7 - 2` | `5` |
| `*` | Multiplication | `7 * 2` | `14` |
| `/` | Division (**always float**) | `7 / 2` | `3.5` |
| `//` | Floor division | `7 // 2` | `3` |
| `%` | Modulus (remainder) | `7 % 2` | `1` |
| `**` | Exponent (power) | `7 ** 2` | `49` |

````python
a, b = 7, 2
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a // b =", a // b)
print("a % b =", a % b)
print("a ** b =", a ** b)
````

### `/` always gives a float

````python
print(8 / 2, type(8 / 2))
````

### Floor division `//` rounds **down** (towards negative infinity)

This matters with negative numbers.

````python
print(7 // 2)     # 3
print(-7 // 2)    # -4  (not -3!) → 3.5 rounded DOWN is -4
print(7.0 // 2)   # 3.0 → a float stays a float
````

### Modulus `%`: the remainder

Common uses: checking even/odd, divisibility, and wrapping around (clock arithmetic).

````python
num = 17
print(num % 2 == 0)     # is it even?
print(num % 5)          # remainder when divided by 5
print((10 + 5) % 12)    # 10 o'clock + 5 hours → 3 o'clock
print(divmod(17, 5))    # (quotient, remainder) in one step
````

### Exponent `**`

````python
print(2 ** 10)     # 1024
print(9 ** 0.5)    # square root → 3.0
print(2 ** -1)     # 0.5
````

### Arithmetic operators work differently with other types

The operator's behaviour depends on the **type** of the operands.

````python
print("ab" + "cd")       # string concatenation
print("ab" * 3)          # string repetition
print([1, 2] + [3])      # list concatenation
print([0] * 4)           # list repetition
````

**⚠️ Error demo:** mixing types that don't support the operator.

````python
# print("5" + 5)         # TypeError
# print("abc" - "a")     # TypeError: strings don't support -
print("5" + str(5), int("5") + 5)
````

### Mixing int and float → result is float

````python
print(3 + 2.0, type(3 + 2.0))
print(True + 2)    # bool acts as 1 → 3
````

## 4. Comparison (Relational) Operators

These compare two values and always return a **`bool`** (`True` or `False`).

| Operator | Meaning | Example | Result |
|----------|---------|---------|--------|
| `==` | Equal to | `5 == 5` | `True` |
| `!=` | Not equal to | `5 != 3` | `True` |
| `>` | Greater than | `5 > 3` | `True` |
| `<` | Less than | `5 < 3` | `False` |
| `>=` | Greater than or equal | `5 >= 5` | `True` |
| `<=` | Less than or equal | `3 <= 2` | `False` |

> ⚠️ `=` assigns a value. `==` compares two values.

````python
x, y = 10, 20
print("Is x > y? :", x > y)
print("Is x < y? :", x < y)
print("Is x == y? :", x == y)
print("Is x >= y? :", x >= y)
print("Is x <= y? :", x <= y)
print("Is x != y? :", x != y)
````

### Chained comparisons (a Python special)

````python
age = 25
print(18 <= age < 60)       # same as: 18 <= age and age < 60
print(1 < 2 < 3 < 4)
````

### Comparing strings: character by character (by Unicode code)

````python
print("apple" < "banana")   # 'a' comes before 'b'
print("Zebra" < "apple")    # True! Capital letters come before lowercase
print(ord("Z"), ord("a"))   # their Unicode numbers
print("10" < "9")           # True! Compares '1' with '9', not the numbers
````

### Comparing floats: be careful with `==`

````python
import math
print(0.1 + 0.2 == 0.3)                 # False (approximate storage)
print(math.isclose(0.1 + 0.2, 0.3))     # True  ✅ the safe way
````

**⚠️ Error demo:** comparing incompatible types with `<` or `>`.

````python
# print(5 < "10")    # TypeError
print(5 == "5")      # == is allowed: just False
````

## 5. Logical Operators

Combine conditions.

| Operator | Returns True when... | Example |
|----------|---------------------|---------|
| `and` | **both** sides are true | `age > 18 and has_id` |
| `or` | **at least one** side is true | `is_admin or is_owner` |
| `not` | it **reverses** the value | `not is_banned` |

### Truth tables

| A | B | A and B | A or B | not A |
|---|---|:-------:|:------:|:-----:|
| True | True | True | True | False |
| True | False | False | True | False |
| False | True | False | True | True |
| False | False | False | False | True |

````python
age = 22
has_id = True
print(age >= 18 and has_id)
print( age < 13 or age > 65)
print(not has_id)
````

### Short-circuit evaluation

Python stops as soon as the answer is known:
- `and` stops at the **first falsy** value
- `or` stops at the **first truthy** value

````python
def check(label, value):
    print("  evaluating", label)
    return value

print("and:")
print(check("left", False) and check("right", True))   # right is never evaluated

print("or:")
print(check("left", True) or check("right", False))    # right is never evaluated
````

### `and` / `or` return one of the operands, not always True/False

````python
print(0 or "default")      # 'default' → first truthy value
print("Ravi" or "default") # 'Ravi'
print(5 and 10)            # 10 → all truthy, so the last value
print(0 and 10)            # 0  → first falsy value

name = ""
display = name or "Guest"  # common pattern: fallback value
print(display)
````

## 6. Assignment Operators

| Operator | Same as |
|----------|---------|
| `x = 5` | assign |
| `x += 3` | `x = x + 3` |
| `x -= 3` | `x = x - 3` |
| `x *= 3` | `x = x * 3` |
| `x /= 3` | `x = x / 3` |
| `x //= 3` | `x = x // 3` |
| `x %= 3` | `x = x % 3` |
| `x **= 3` | `x = x ** 3` |
| `x &= 3`, `x \|= 3`, `x ^= 3`, `x >>= 3`, `x <<= 3` | bitwise versions |

> Python has **no** `x++` or `x--`.

````python
x = 10
x += 5;  print(x)
x -= 3;  print(x)
x *= 2;  print(x)
x /= 4;  print(x)
x //= 2; print(x)
x **= 2; print(x)
x %= 5;  print(x)
````

### Walrus operator `:=` (Python 3.8+)

It assigns a value **and** returns it, all inside an expression.

````python
nums = [3, 8, 1, 9, 4]
if (n := len(nums)) > 3:
    print(f"List is long: {n} items")
````

## 7. Identity Operators: `is`, `is not`

These check whether two names refer to the **same object in memory** (same `id()`). They don't compare values. That's what `==` is for.

````python
a = [1, 2, 3]
b = a
c = [1, 2, 3]

print(a == c)        # True  → same values
print(a is c)        # False → different objects
print(a is b)        # True  → same object
print(id(a), id(b), id(c))
````

> ✅ The main real-world use of `is` is checking for `None`:

````python
result = None
print(result is None)
print(result is not None)
````

## 8. Membership Operators: `in`, `not in`

These check whether a value exists **inside** a collection or string.

````python
fruits = ["apple", "banana", "mango"]
print("apple" in fruits)
print("kiwi" not in fruits)

print("Py" in "Python")              # substring check
print("e" in "hello")

student = {"name": "Ravi", "age": 25}
print("name" in student)             # dicts: checks KEYS
print("Ravi" in student)             # False → not a key
print("Ravi" in student.values())    # True
````

## 9. Bitwise Operators

These work on the **binary (bit) form** of integers.

| Operator | Name | Example (`a=5` → `0101`, `b=3` → `0011`) | Result |
|----------|------|---------|--------|
| `&` | AND | `5 & 3` → `0001` | `1` |
| `\|` | OR | `5 \| 3` → `0111` | `7` |
| `^` | XOR | `5 ^ 3` → `0110` | `6` |
| `~` | NOT | `~5` | `-6` (equals `-(x+1)`) |
| `<<` | Left shift | `5 << 1` → `1010` | `10` (× 2) |
| `>>` | Right shift | `5 >> 1` → `0010` | `2` (÷ 2) |

````python
a, b = 5, 3
print(bin(a), bin(b))
print("a & b  =", a & b, bin(a & b))
print("a | b  =", a | b, bin(a | b))
print("a ^ b  =", a ^ b, bin(a ^ b))
print("~a     =", ~a)
print("a << 1 =", a << 1)
print("a >> 1 =", a >> 1)
````

## 10. Conditional (Ternary) Expression

A one-line `if-else` that **produces a value**:

```
value_if_true  if  condition  else  value_if_false
```

````python
age = 20
status = "Adult" if age >= 18 else "Minor"
print(status)

n = 7
print("Even" if n % 2 == 0 else "Odd")
````

## 11. Operator Precedence (Order of Evaluation)

When an expression has several operators, Python follows this order, from **highest to lowest**:

| Priority | Operators | Description |
|:-:|-----------|-------------|
| 1 | `()` | Parentheses |
| 2 | `**` | Exponent |
| 3 | `+x` `-x` `~x` | Unary plus, minus, bitwise NOT |
| 4 | `*` `/` `//` `%` | Multiply, divide, floor divide, modulus |
| 5 | `+` `-` | Add, subtract |
| 6 | `<<` `>>` | Bit shifts |
| 7 | `&` | Bitwise AND |
| 8 | `^` | Bitwise XOR |
| 9 | `\|` | Bitwise OR |
| 10 | `==` `!=` `>` `<` `>=` `<=` `is` `is not` `in` `not in` | Comparisons, identity, membership |
| 11 | `not` | Logical NOT |
| 12 | `and` | Logical AND |
| 13 | `or` | Logical OR |
| 14 | `if – else` | Conditional expression |
| 15 | `:=` | Walrus (assignment expression) |

Memory aid for arithmetic: **PEMDAS**, which stands for **P**arentheses, **E**xponents, **M**ultiplication/**D**ivision, **A**ddition/**S**ubtraction.

````python
print(2 + 3 * 4)        # 14, not 20 → * before +
print((2 + 3) * 4)      # 20 → parentheses first
print(10 - 4 - 3)       # 3 → left to right
print(2 * 3 ** 2)       # 18 → ** before *
````

### Associativity: same priority, which side goes first?

- Most operators go **left → right**
- `**` goes **right → left**

````python
print(100 / 10 / 5)     # (100 / 10) / 5 = 2.0
print(2 ** 3 ** 2)      # 2 ** (3 ** 2) = 2 ** 9 = 512, not 64
````

````python
print(-2 ** 2)          # -4 → ** comes before unary minus: -(2**2)
print((-2) ** 2)        # 4
print(not 5 > 3)        # False → comparison first, then not
print(True or False and False)   # True → 'and' before 'or'
````

> ✅ **Best practice:** when in doubt, **use parentheses**. They make code clearer even when they aren't strictly needed.

## 12. Step-by-Step Evaluation Example

```
result = 10 + 2 * 3 ** 2 > 20 and not False
```

| Step | Rule applied | Expression becomes |
|:-:|------|------------|
| 1 | `**` first | `10 + 2 * 9 > 20 and not False` |
| 2 | `*` | `10 + 18 > 20 and not False` |
| 3 | `+` | `28 > 20 and not False` |
| 4 | comparison | `True and not False` |
| 5 | `not` | `True and True` |
| 6 | `and` | `True` |

## 13. Useful Built-in Functions for Expressions

````python
print(abs(-7))           # absolute value → 7
print(round(3.14159, 2)) # 3.14
print(round(2.5))        # 2  ← rounds to the nearest EVEN number ("banker's rounding")
print(round(3.5))        # 4
print(max(4, 9, 2), min(4, 9, 2))
print(pow(2, 5))         # same as 2 ** 5
print(divmod(17, 5))     # (3, 2)
````
