# Part 1: Introduction to Control Flow

## 1. What is Control Flow?

**Control flow** is the **order in which a program's statements are executed**.

By default, Python runs code **top to bottom, one line at a time**. Control flow statements let you **change that order**, so you can:
- **make decisions**: run code only if a condition is true
- **repeat** code many times
- **jump**: skip or stop code early

> Without control flow, a program would do exactly the same thing every time. With it, a program can **react** to data, to user input, and to different situations.

## 2. The Three Basic Structures

Every program, in any language, is built from these three building blocks:

| Structure | Meaning | Python keywords |
|-----------|---------|-----------------|
| **1. Sequential** | Run statements one after another | *(default, no keyword)* |
| **2. Selection (Decision / Conditional)** | Choose a path based on a condition | `if`, `elif`, `else`, `match`-`case` |
| **3. Repetition (Iteration / Loops)** | Repeat a block of code | `for`, `while` |

Plus **jump / transfer statements**, which change the flow inside the above:

| Statement | Purpose |
|-----------|---------|
| `break` | exit a loop immediately |
| `continue` | skip to the next loop iteration |
| `pass` | do nothing (a placeholder) |
| `return` | exit a function and send back a value |

```
   SEQUENTIAL            SELECTION                 REPETITION
   ┌────────┐           ┌──────────┐              ┌──────────┐
   │ step 1 │           │condition?│◄─────┐       │condition?│──No──► exit
   └───┬────┘           └──┬────┬──┘      │       └────┬─────┘
       ▼              True │    │ False   │        True│
   ┌────────┐         ┌────▼─┐ ┌▼─────┐   │       ┌────▼─────┐
   │ step 2 │         │ do A │ │ do B │   │       │  body    │
   └───┬────┘         └────┬─┘ └┬─────┘   │       └────┬─────┘
       ▼                   └─┬──┘         └────────────┘
   ┌────────┐                ▼
   │ step 3 │            continue
   └────────┘
```

This notebook covers **sequential** flow and **selection** (`if`). Loops get their own notebook.

## 4. Indentation and Code Blocks (very important in Python)

Other languages use `{ }` to group code. **Python uses indentation** (spaces at the start of a line).

```python
if condition:          # ← the header line ends with a colon  :
    statement_1        # ← indented: this is INSIDE the block
    statement_2        # ← same indentation: also inside
statement_3            # ← back to the left: OUTSIDE the block, always runs
```

Rules:
- The header line (`if`, `elif`, `else`, `for`, `def`...) **must end with `:`**
- The block underneath **must be indented**. The standard is **4 spaces** (Jupyter does this automatically when you press Tab or Enter after `:`)
- All lines in the same block need the **same indentation**
- Never mix tabs and spaces

````python
temperature = 35

if temperature > 30:
    print("It's hot!")           # inside the if
    print("Drink water.")        # inside the if
print("This always runs.")       # outside the if
````

# Part 2: Conditional Statements (`if` / `elif` / `else`)

## 5. The `if` Statement

It runs a block **only if** the condition is `True`.

```python
if condition:
    # runs only when condition is True
```

```
        ┌────────────┐
        │ condition? │
        └──┬──────┬──┘
      True │      │ False
     ┌─────▼────┐ │
     │  block   │ │
     └─────┬────┘ │
           └──┬───┘
              ▼
          next line
```

````python
age = 20

if age >= 18:
    print("You are an adult.")

print("Program continues...")
````

Change `age` to `15` and run again. The message inside the `if` is skipped.

### The condition can be any expression
Python checks whether its value is **truthy** or **falsy** (see the data types notebook).

````python
age = 15

if age >= 18:
    print("You are an adult.")

print("Program continues...")
````

````python
name = "Ravi"
cart = []

if name:                      # non-empty string → truthy
    print(f"Hello, {name}!")

if cart:                      # empty list → falsy, so this is skipped
    print("You have items in your cart.")

if not cart:
    print("Your cart is empty.")
````

## 6. The `if-else` Statement

Choose between **two** paths: one runs if the condition is True, the other if it's False.
**Exactly one** of the two blocks always runs.

```python
if condition:
    # runs when True
else:
    # runs when False
```

```
        ┌────────────┐
        │ condition? │
        └──┬──────┬──┘
      True │      │ False
    ┌──────▼─┐  ┌─▼──────┐
    │ block A│  │ block B│
    └──────┬─┘  └─┬──────┘
           └──┬───┘
              ▼
```

````python
number = 7

if number % 2 == 0:
    print(f"{number} is even")
else:
    print(f"{number} is odd")
````

````python
number = 10

if number % 2 == 0:
    print(f"{number} is even")
else:
    print(f"{number} is odd")
````

## 7. The `if-elif-else` Chain

Use it for **more than two** choices. `elif` means "else if".

```python
if condition_1:
    # runs if condition_1 is True
elif condition_2:
    # runs if condition_1 is False AND condition_2 is True
elif condition_3:
    # ...
else:
    # runs if NONE of the above are True
```

**How it works:**
- Python checks the conditions **from top to bottom**
- It runs the **first** block whose condition is True, then **skips all the rest**
- `else` is optional and catches everything else
- You can have as many `elif`s as you need

````python
marks = 72
if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >=60:
    grade = "C"
elif marks >= 40:
    grade = "D"
else:
    grade = "F"
print(f"{marks} -> {grade}")
````

### ⚠️ Order matters!

Since only the **first** True branch runs, put the **most specific** (or highest) condition first.

````python
marks = 95

# ❌ Wrong order: 95 >= 40 is True first, so it never reaches the A check
if marks >= 40:
    print("Wrong:", "D")
elif marks >= 90:
    print("Wrong:", "A")

# ✅ Correct order
if marks >= 90:
    print("Correct:", "A")
elif marks >= 40:
    print("Correct:", "D")
````

### `if-elif` vs several separate `if`s

````python
x = 15

print("Using if-elif (only the first match runs):")
if x > 10:
    print("  greater than 10")
elif x > 5:
    print("  greater than 5")

print("Using separate ifs (every condition is checked):")
if x > 10:
    print("  greater than 10")
if x > 5:
    print("  greater than 5")
````

## 8. Combining Conditions with `and`, `or`, `not`

````python
age = 25
has_license = True
if age >= 18 and has_license:
    print("You can drive.")
elif age>=18 and not has_license:
    print("Get a License First")
else:
    print("Too young to drive.")
````

````python
day = "Sunday"

if day == "Saturday" or day == "Sunday":
    print("Weekend!")
else:
    print("Weekday.")

# Cleaner, using membership:
if day in ("Saturday", "Sunday"):
    print("Weekend! (using in)")
````

````python
# Chained comparison for ranges
temp = 22

if 18 <= temp <= 26:
    print("Comfortable temperature")
````

## 9. Nested `if` Statements

An `if` inside another `if`. Use it when a second decision only makes sense **after** the first one.

````python
username = "admin"
password = "1234"

if username == "admin":
    if password == "1234":
        print("Login Successful")
    else:
        print("Wrong Password")
else:
    print("Unknown User")
````

> ⚠️ Too much nesting makes code hard to read. Often you can **flatten** it with `and` or `elif`:

````python
username = "admin"
password = "1234"

if username != "admin":
    print("Unknown user")
elif password != "1234":
    print("Wrong password")
else:
    print("Login successful")
````

## 10. Conditional (Ternary) Expression: a one-line if-else

```python
value = value_if_true if condition else value_if_false
```

Use it for **simple choices between two values**. Use a normal `if-else` for anything with several steps.

````python
age = 16
status = "Adult" if age >= 18 else "Minor"
print(status)
````

````python
n = -5
print("Positive" if n > 0 else "zero" if n == 0 else "Negative")
````

## 12. `match`-`case` (Python 3.10+): a quick look

When you compare **one value** against many fixed options, `match` can be cleaner than a long `elif` chain.
(It can do much more, called *pattern matching*, which is covered later.)

````python
command = "stop"

match command:
    case "start":
        print("Starting...")
    case "stop":
        print("Stopping...")
    case "pause" | "wait":          # | means OR
        print("Pausing...")
    case _:                         # _ = default, like else
        print("Unknown command")
````

## 14. Common Mistakes

| Mistake | Problem | Fix |
|---------|---------|-----|
| `if x = 5:` | `=` is assignment → SyntaxError | `if x == 5:` |
| `if x > 5` | Missing colon → SyntaxError | `if x > 5:` |
| Block not indented | IndentationError | Indent with 4 spaces |
| `else if x > 5:` | Not Python syntax | `elif x > 5:` |
| `else x < 5:` | `else` takes no condition | `else:` or `elif x < 5:` |
| `if color == "red" or "blue":` | Always True | `if color in ("red", "blue"):` |
| Wrong `elif` order | A broad condition catches everything first | Put specific / highest checks first |
| `if input("Age: ") > 18:` | Comparing str with int → TypeError | `int(input(...))` |

## 15. Key Takeaways

1. **Control flow** decides the **order** in which code runs: **sequential**, **selection** and **repetition**.
2. Python uses **indentation (4 spaces)** and a **colon `:`** to define blocks.
3. **`if`**: run a block only when the condition is True.
4. **`if-else`**: choose between two paths. Exactly one runs.
5. **`if-elif-else`**: many paths. Only the **first** True branch runs, so **order matters**.
6. Conditions can be any expression, evaluated as **truthy or falsy**.
7. Combine conditions with `and` / `or` / `not`, `in`, and chained comparisons.
8. Avoid deep nesting. Flatten it with `elif` or `and`.
9. `pass` is a placeholder. The ternary expression is a one-line if-else. `match`-`case` handles many fixed options.
