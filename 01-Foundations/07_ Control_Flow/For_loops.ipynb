## Part 1 — Why Loops Exist 🟢

A **loop** repeats a block of code. Without loops, printing 1 to 5 looks like this:

````python
print(1)
print(2)
print(3)
print(4)
````

Now imagine 1 to 1,000,000. Loops let you write the instruction **once** and repeat it.

Python has two loop statements, `for` and `while`. This notebook is all about **`for`**:
use it when you want to go **through each item** of a collection or sequence — "**For** every item in this…".

```
        ┌───────────────────────┐
        │  any items left?      │◄──────────┐
        └──────────┬────────────┘           │
           Yes     │     No ──► exit loop   │
                   ▼                        │
        ┌───────────────────────┐           │
        │ item = next item      │           │
        │ run the loop body     │───────────┘
        └───────────────────────┘
```

````python
for n in range(1,6):
    print(n)
````

## Part 2 — The `for` Loop 🟢

### 2.1 Syntax
```python
for item in iterable:
    # body runs once per item
```
An **iterable** is anything you can loop over: `str`, `list`, `tuple`, `dict`, `set`, `range`, files, and more.

> Python's `for` is a **for-each** loop. It is *not* like C's `for (i=0; i<n; i++)` — it hands you the items directly.

````python
fruits = ["apple", "banana", "cherry"]
for item in fruits:
    print("I like", item)
````

### 2.2 Looping over different iterables

````python
# string -> characters
for char in "Python":
    print(char, end = " | ")
````

````python
# Tuple
for x in (10, 20, 30):
    print(x * 2, end = " ")
````

````python
# Set
for colour in {"red", "green", "blue"}:
    print(colour, end = " ")
````

### 2.3 The loop variable survives the loop

````python
for i in [3, 6, 9]:
    pass
print("After the loop, i =", i)
````

## Part 3 — `range()` in Depth 🟢

`range(start, stop, step)` produces integers from `start` **up to but not including** `stop`.

| Call | Produces |
|---|---|
| `range(5)` | 0, 1, 2, 3, 4 |
| `range(2, 6)` | 2, 3, 4, 5 |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 |
| `range(10, 0, -1)` | 10, 9, …, 1 |
| `range(5, 2)` | *(empty)* |

````python
print(list(range(5)))
print(list(range(2,6)))
print(list(range(0, 10, 2)))
print(list(range(10, 0, -1)))
print(list(range(5, 2)))
````

### 3.1 `range` is lazy (memory-efficient)
`range` does **not** build a list. It stores only `start`, `stop`, `step` and calculates each number on demand.

````python
import sys
r = range(10_00_000)
l = list(range(1000))
print("range(10 million) uses", sys.getsizeof(r), "bytes")
print("list of only 1000 ints uses", sys.getsizeof(l), "bytes")
print("Indexing works:", r[5_00_000], "| membership is O(1):", 9_99_999 in r)
print("Indexing works:", r[5_000_000], "| membership is O(1):", 9_999_999 in r)
````

### 3.2 Practical `range` examples

````python
# Multiplcation table
n = 7
for i in range(1,11):
    print(f"{n} * {i:2} = {n * i:3}")
````

````python
# Sum of 1..100 and sum of even numbers 1..100
print("Sum 1..100       :", sum(range(1, 101)))
print("Sum evens 1..100 :", sum(range(2, 101,2)))
````

````python
# Factorial with a loop
n, fact = 6, 1
for k in range(2, n+1):
    fact *= k
print(f"{n}! =", fact)
````

````python
# Fibonacci series
a, b = 0, 1
for _ in range(12):
    print(a, end = " ")
    a, b = b, a+b
````

## Part 4 — `break`, `continue`, `pass`, and the Loop `else` 🟡

| Keyword | What it does |
|---|---|
| `break` | **Exit** the loop immediately |
| `continue` | **Skip** the rest of this iteration, go to the next one |
| `pass` | Do **nothing** (placeholder for empty blocks) |
| `else` (on a loop) | Runs **only if the loop finished without `break`** |

````python
# break: stop at the first negative number
for n in [4, 8, -1, 15, 16]:
    if n < 0:
        print("Found something negative, stopping")
        break
    print("Processing", n)
````

````python
# continue: skip odd numbers
for n in range(1, 11):
    if n % 2 != 0:
        continue
    print(n, end = " ")
````

````python
# pass: a placeholder — the code must be syntactically complete
for n in range(3):
    pass          # TODO: implement later
print("pass did nothing, loop ran fine")
````

### 4.1 The loop `else` clause (unique to Python)
Think of it as **"no break"**. Most useful for *search* loops.

````python
n = 97        # try changing this to 97, 17, 100 and re-run

for d in range(2, n):
    if n % d ==0:
        print(f"{n} is not prime becuae it is divisible by {d}")
        break
else:
    print(f"{n} is prime.")
    
````

## Part 5 — Nested Loops & Patterns 🟡

A loop inside a loop. The **inner loop runs completely for every single iteration of the outer loop.**

If outer runs `m` times and inner runs `n` times → body runs `m × n` times.

````python
for row in range(1, 4):
    for col in range(1,4):
        print(f"({row},{col})", end = " ")
````

````python
for row in range(1, 4):
    for col in range(1,4):
        print(f"({row},{col})", end = " ")
    print()
````

````python
n = 5
for i in range (1, n+1):
    print("* " * i)
````

````python
n = 5
for i in range (n, 0, -1):
    print("* " * i)
````

````python
n = 5
for i in range (1, n+1):
    print(" " * (n - i) + "* " * i)
````

````python
n = 5
for i in range (n, 0, -1):
    print(" " * (n - i) + "* " * i)
````

````python
n = 5
for i in range (1, n):
    print(" " * (n - i) + "* " * i)
for i in range (n, 0, -1):
    print(" " * (n - i) + "* " * i)
````

````python
# Number patterns using a true nested loop
n = 5
for i in range(1, n+1):
    for j in range(1, i+1):
        print(j, end = " ")
    print()
````

````python
for i in range(1, 10):
    for j in range(1, 10):
        print(f"{i * j:3}", end ='')
    print()
````

### 5.2 Breaking out of nested loops
`break` only exits the **innermost** loop. Two ways to exit both:

````python
grid = [[1, 2, 3], [4, 99, 6], [7, 8, 9]]
target = 99

# Way 1: a flag
found = False
for r, row in enumerate(grid):
    for c, val in enumerate(row):
        if val == target:
            found = True
            break
    if found:
        break
print("Flag method   :", (r, c))
````

````python

# Way 2: for-else trick
for r, row in enumerate(grid):
    for c, val in enumerate(row):
        if val == target:
            break
    else:
        continue      # inner loop didn't break → keep going
    break             # inner loop did break → break outer too
print("for-else trick:", (r, c))
````

## Part 6 — Built-in Loop Helpers 🟡

### 6.1 `enumerate()` — get index **and** value
❌ Beginner style: `for i in range(len(lst)): print(i, lst[i])`
✅ Pythonic style: `for i, item in enumerate(lst)`

````python
students = ["Asha", "Ben", "Chen"]
for i, name in enumerate(students):
    print(i, name)
````

````python
for rank, name in enumerate(students, start = 1):
    print(f"Rank {rank}: {name}")
````

````python
names  = ["Asha", "Ben", "Chen"]
scores = [91, 78, 85]
cities = ["Pune", "Leeds", "Austin"]

for names, scores, cities in zip(names, scores, cities):
    print(f"{names:5}, {scores}, {cities:6}")
````

````python

# zip stops at the SHORTEST input
print(list(zip([1, 2, 3], ["a", "b"])))
````

````python
nums = [5, 2, 9, 1]
for n in reversed(nums):
    print(n, end = " ")
print("  ← reversed (original unchanged:", nums, ")")
````

````python
for n in sorted(nums):
    print(n, end=" ")
print("  ← sorted")
````

````python
words = ["banana", "Apple", "cherry", "kiwi"]
for w in sorted(words, key=len, reverse=True):
    print(w, end = " ")
print("  ← sorted by length, longest first")
````

## Part 7 — Classic Loop Patterns 🟡
Nearly every loop you'll ever write is one of these patterns.

````python
data = [23, 7, 42, 15, 8, 42, 4, 16]

# 1. Accumulator (sum / product)
total = 0
for x in data:
    total += x
print("1. Sum           :", total)

# 2. Counter
evens = 0
for x in data:
    if x % 2 == 0:
        evens += 1
print("2. Even count    :", evens)

# 3. Max / min tracker
biggest = data[0]
for x in data[1:]:
    if x > biggest:
        biggest = x
print("3. Max           :", biggest)

# 4. Linear search (first index)
target, pos = 42, -1
for i, x in enumerate(data):
    if x == target:
        pos = i
        break
print("4. First 42 at   :", pos)

# 5. Filter (build a new list)
big = []
for x in data:
    if x > 10:
        big.append(x)
print("5. Values > 10   :", big)

# 6. Map / transform
squares = []
for x in data:
    squares.append(x * x)
print("6. Squares       :", squares)

# 7. Flag (any / all)
has_negative = False
for x in data:
    if x < 0:
        has_negative = True
        break
print("7. Any negative? :", has_negative)

# 8. Frequency count
freq = {}
for x in data:
    freq[x] = freq.get(x, 0) + 1
print("8. Frequencies   :", freq)
````

### 8.1 Built-ins that replace these loops
Once you understand the patterns, use the built-ins — they are shorter and faster (they run in C).

````python
from collections import Counter
data = [23, 7, 42, 15, 8, 42, 4, 16]
print("sum   :", sum(data))
print("max   :", max(data), "| min:", min(data))
print("count :", data.count(42), "← how many times 42 appears")
print("index :", data.index(42))
print("freq  :", Counter(data))
````

## Part 9 — Common Mistakes ⚠️

| # | Mistake | Symptom | Fix |
|---|---|---|---|
| 1 | Off-by-one with `range` | Missing last value | `range(1, n + 1)` |
| 2 | Modifying a list/dict while looping over it | Skipped items / `RuntimeError` | Loop over a copy, or build a new collection |
| 3 | `for i in range(len(lst))` everywhere | Works, but un-Pythonic | `enumerate(lst)` |
| 4 | Using `=` instead of `==` in conditions | `SyntaxError` | `==` compares |
| 5 | Wrong indentation | Code runs outside/inside loop by accident | Check 4-space alignment |
| 6 | Changing the loop variable to "skip" items | Has no effect | Use `continue` |
| 7 | Forgetting the colon `:` after `for ... in ...` | `SyntaxError` | `for x in items:` |

````python
# Mistake 5: indentation changes meaning
total = 0
for n in [1, 2, 3]:
    total += n
    print("inside :", total)     # runs 3 times
print("outside:", total)         # runs once
````

````python
# Mistake 6: assigning to the loop variable doesn't affect the next iteration
for i in range(5):
    print(i, end=" ")
    i += 10          # ignored: range hands out the next value regardless
````

## Part 11 — Cheat Sheet 📋

```python
# ── for ─────────────────────────────────────────────
for x in iterable: ...
for i in range(start, stop, step): ...
for i, x in enumerate(seq, start=0): ...
for a, b in zip(seq1, seq2, strict=False): ...
for k, v in d.items(): ...
for x in reversed(seq): ...
for x in sorted(seq, key=..., reverse=...): ...

# ── control ─────────────────────────────────────────
break        # leave loop
continue     # next iteration
pass         # do nothing
for ...: ... else: ...   # else runs if no break
```

### 🧠 Which `for` style?
```
Looping over…
├── items only?            → for x in items
├── items + their index?   → for i, x in enumerate(items)
├── two lists together?    → for a, b in zip(list1, list2)
├── a dictionary?          → for k, v in d.items()
└── a count of numbers?    → for i in range(n)
```
