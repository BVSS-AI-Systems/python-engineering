## Part 1 — Why `while` Loops Exist 🟢

A `for` loop goes through a collection: **you know what you're looping over**.

A `while` loop keeps going **as long as a condition is `True`**. Use it when **you don't know in advance how many times** the loop will run:

- Keep asking for a password **until** it's correct
- Keep dividing a number by 10 **until** it reaches 0
- Keep a game running **until** the player quits
- Keep saving money **until** you reach a target

```
        ┌──────────────────┐
        │   condition?     │◄──────────┐
        └────────┬─────────┘           │
          True   │   False ──► exit    │
                 ▼                     │
        ┌──────────────────┐           │
        │   loop body      │───────────┘
        │ (must change     │
        │  something!)     │
        └──────────────────┘
```

**Read it as:** "*While* this is true, keep doing this."

````python
balance = 100
while balance > 0:
    print("Spending 30...balance was", balance)
    balance -= 30
print("Out of money! Final balance:", balance)
````

## Part 2 — Syntax, the 3 Parts, and Tracing 🟢

### 2.1 Syntax
```python
while condition:
    # body (indented 4 spaces)
    # runs again and again while condition is True
# code here runs after the loop ends
```

### 2.2 The three parts every `while` loop needs
| Part | Where | Example |
|---|---|---|
| 1. **Initialise** | before the loop | `count = 1` |
| 2. **Check** | in the condition | `while count <= 5:` |
| 3. **Update** | inside the body | `count += 1` |

Forget part 3 and the loop never ends (an **infinite loop**).

````python
count = 1            # 1. initialise
while count <= 5:    # 2. check
    print("count =", count)
    count += 1       # 3. update  (same as count = count + 1)

print("Loop finished, count is now", count)
````

### 2.3 Tracing a loop by hand (dry run)
Good programmers *trace* loops on paper. For the cell above:

| Iteration | `count` before check | `count <= 5`? | Printed | `count` after update |
|---|---|---|---|---|
| 1 | 1 | ✅ | count = 1 | 2 |
| 2 | 2 | ✅ | count = 2 | 3 |
| 3 | 3 | ✅ | count = 3 | 4 |
| 4 | 4 | ✅ | count = 4 | 5 |
| 5 | 5 | ✅ | count = 5 | 6 |
| — | 6 | ❌ exit | — | — |

Notice: after the loop, `count` is **6**, not 5. The condition is checked **before** every iteration, including one last time when it fails.

````python
x = 10
while x < 5:
    print("This never prints")
    x += 1
print("The body ran 0 times because 10 < 5 was False from the start")
````

### 2.5 Any expression that is "truthy" or "falsy" works as a condition
`0`, `""` (empty string), `[]` (empty list) and `None` count as **False**. Everything else counts as **True**.

````python
n = 3
while n:
    print("n =", n)
    n-=1
````

````python
word = "abc"
while word:
    print(word)
    word = word[1:]
````

## Part 3 — Counting Loops 🟢

````python
# Count up
i = 1
while i <= 10:
    print(i, end=" ")
    i += 1
print()

# Count down
i = 10
while i >= 1:
    print(i, end=" ")
    i -= 1
print("Liftoff! 🚀")

# Count in steps
i = 0
while i <= 50:
    print(i, end=" ")
    i += 5
print()
````

````python
n = 7
i = 1
while i <= 10:
    print(f"{n} x {i:2} = {n*i:3}")
    i+=1
````

````python
# sum 1 to 100
n = 1
sum = 0
while n<=100:
    sum+=n
    n+=1
print("Sum of 1 to 100 numbers is :", sum)
````

````python
# factorial of 6
n, fact = 6, 1
i=2
if n <0:
    print("No factorial for negative numbers")
elif n==0 or n==1:
    print(f"The factorial of {n} is 1")
else:
    while i<=n:
        fact *= i
        i += 1
print(f"{n}! = {fact}")
````

````python
a, b = 0, 1
while a < 200:
    print(a, end = " ")
    a, b = b, a+b
````

````python
# Powers of 2 up to 1000
p = 1
while p <= 1000:
    print(p, end = " ")
    p*=2
````

## Part 4 — Working with Numbers Digit by Digit 🟢

Two operators do all the work:
| Operator | Meaning | `9876 ...` |
|---|---|---|
| `% 10` | last digit | `9876 % 10 → 6` |
| `// 10` | drop the last digit | `9876 // 10 → 987` |

We don't know how many digits a number has, so `while num > 0` is perfect.

````python
num = 98765
digits_seen = []
while num > 0:
    last = num % 10
    digits_seen.append(last)
    print(f"num = {num:5} -> last digit {last}, remaining {num//10}")
    num //=10
print("Digits (right to left):", digits_seen)
````

````python
# Sum of digits
num, total = 98765, 0
while num > 0:
    total += num % 10
    num //= 10
print("Sum of digits:", total)

# Count digits
num, count = 1234567, 0
while num > 0:
    count += 1
    num //= 10
print("Number of digits:", count)

# Reverse a number
num, rev = 12345, 0
while num > 0:
    rev = rev * 10 + num % 10
    num //= 10
print("Reversed:", rev)
````

### 4.1 GCD with Euclid's algorithm (2,300 years old!)
Replace `(a, b)` with `(b, a % b)` until `b` becomes 0. Then `a` is the greatest common divisor.

````python
a, b = 84, 36
print(f"GCD of {a} and {b}:")
while b != 0:
    print(f"  a = {a:3}, b = {b:3}, a % b = {a % b}")
    a, b = b, a % b
print("  GCD =", a)
````

### 4.2 Decimal to binary
Divide by 2 repeatedly; the remainders (read backwards) are the binary digits.

````python
n = 45
original = n
binary = ""
while n > 0:
    binary = str(n % 2) + binary
    n //= 2
print(f"{original} in binary is {binary}  (check: {bin(original)})")
````

## Part 5 — Sentinel Loops 🟡

A **sentinel** is a special value that means "stop". The loop keeps reading values until it sees the sentinel.

**Pattern:**
```python
value = read_first()
while value != SENTINEL:
    process(value)
    value = read_next()      # read again at the END of the body
```

````python
fake_inputs = iter(["12", "7", "30", "done", "99"])   # pretend the user typed these

total = 0
count = 0
entry = next(fake_inputs)
while entry!="done":
    total+= int(entry)
    count+=1
    entry = next(fake_inputs)
print(f"You entered {count} numbers, total = {total}, average = {total / count:.2f}")    
````

````python
# Sentinel = a negative number. Track the highest score.
scores = iter(["67", "92", "81", "-1"])

highest = None
score = int(next(scores))          # real code: int(input("Score (-1 to stop): "))
while score >= 0:
    if highest is None or score > highest:
        highest = score
    score = int(next(scores))
print("Highest score:", highest)
````

⚠️ **Why read before the loop *and* at the end of the body?** If you only read inside at the top, the sentinel value itself gets processed before the check. Try moving the line and see what happens.

## Part 6 — `while True` + `break` 🟡

`while True` is an **intentional** infinite loop. You exit from the **middle** with `break`. This is the cleanest pattern when you must do something *before* you can decide whether to stop.

````python
attempts = iter(["1234", "abcd", "secret"])
password = "secret"

while True:
    guess = next(attempts)
    if guess == password:
        print("Access Granted")
        break
    print("Wrong guess, try again")
````

````python
answers = iter(["abc", "-5", "150", "42"])

while True:
    raw = next(answers)                 # real code: raw = input("Enter age (0-120): ")
    if not raw.isdigit():
        print(f"'{raw}' is not a positive whole number")
        continue
    age = int(raw)
    if age > 120:
        print(f"{age} is out of range")
        continue
    break
print("Valid age:", age)
````

### 6.3 Limited retries — stop after 3 wrong tries

````python
attempts = iter(["0000", "1111", "2222", "9999"])
PIN = "9999"
tries = 0
MAX_TRIES = 3

while tries < MAX_TRIES:
    pin = next(attempts)
    tries+=1
    if pin == PIN:
        print("Unlocked")
        break
    print(f"❌ Wrong PIN ({MAX_TRIES - tries} tries left)")
else:
    print("🔒 Card blocked")
````

## Part 7 — `break`, `continue`, `pass`, and `while ... else` 🟡

| Keyword | What it does |
|---|---|
| `break` | **Exit** the loop immediately |
| `continue` | **Skip** the rest of this iteration and go back to the condition |
| `pass` | Do **nothing** (placeholder) |
| `else` (on a loop) | Runs **only if the loop ended because the condition became False** (no `break`) |

````python
# break: stop at the first multiple of 7 above 50
n = 51
while True:
    if n % 7 == 0:
        print("First multiple of 7 above 50:", n)
        break
    n += 1
````

````python
# continue: print 1..10 but skip multiples of 3
# ⚠️ update BEFORE continue, or the loop gets stuck!
n = 0
while n < 10:
    n += 1              # update first
    if n % 3 == 0:
        continue
    print(n, end=" ")
````

````python
# pass: placeholder while you're still writing the code
n = 0
while n < 3:
    pass                # TODO
    n += 1
print("Loop ran", n, "times doing nothing")
````

### 7.1 `while ... else` — "no break"
Perfect for **search** loops: the `else` runs only when the search finished without finding anything.

````python
stock = ["pen", "book", "lamp"]

for wanted in ["book", "phone"]:       # (a for loop just to test two values)
    i = 0
    while i < len(stock):
        if stock[i] == wanted:
            print(f"'{wanted}' found at index {i}")
            break
        i += 1
    else:
        print(f"'{wanted}' not in stock")
````

## Part 8 — Nested `while` Loops & Patterns 🟡

The inner loop runs **completely** for every iteration of the outer loop.
⚠️ With nested `while` loops, **reset the inner counter inside the outer loop**.

````python
row =1
while row <=3:
    col = 1
    while col<=4:
        print(f"({row},{col})",end = " ")
        col +=1
    print()
    row += 1
````

````python
n = 1
while n <= 5:
    print("* "*n, end = " ")
    n+=1
    print()
````

````python
n = 5
while n >= 1:
    print("* "*n, end = " ")
    n-=1
    print()
````

````python
n = 5
i = 1
while i <= n:
    print(" "* (n-i) + "* " * i)
    i+=1
    
````

````python
i = 1
n = 5
while i <= n:
    j = 1
    while j <= i:
        print(j, end = " ")
        j += 1
    print()
    i+=1
````

````python
# Multiplication grid 1..9
i = 1
while i <= 9:
    j = 1
    while j <= 9:
        print(f"{i*j:3}", end="")
        j += 1
    print()
    i += 1
````
